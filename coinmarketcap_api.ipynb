```python
# API FUNCTION

import os
import json
import ssl
import urllib.parse
import urllib.request
import certifi
import pandas as pd

def api_runner():
    params = urllib.parse.urlencode({
        "start": "1",
        "limit": "10",
        "convert": "USD",
    })
    
    url = f"https://pro-api.coinmarketcap.com/v1/cryptocurrency/listings/latest?{params}"

    request = urllib.request.Request(
        url,
        headers={
            "Accept": "application/json",
            "X-CMC_PRO_API_KEY": "d2289781ec0c4107ae846be37be60a88",
        },
    )

    context = ssl.create_default_context(cafile=certifi.where())
    with urllib.request.urlopen(request, context=context) as response:
        data = json.load(response)

    df = pd.json_normalize(data['data'])
    
    df['Timestamp'] = pd.Timestamp.now()

    pd.reset_option('display.width')
    pd.reset_option('display.max_columns')
    pd.set_option('display.max_columns', None)
    pd.set_option('display.max_rows', None)

    file_path = r'C:\Users\user\Documents\anaconda\conda\Output\CryptoAPI.csv'
    
    os.makedirs(os.path.dirname(file_path), exist_ok=True)

    if not os.path.isfile(file_path):
        df.to_csv(file_path, index=False, header=True)
        print(f"📁 Created new file and wrote {len(df)} rows!")
    else:
        df.to_csv(file_path, mode='a', index=False, header=False)
        print(f"📝 Appended {len(df)} new rows to historical dataset!")
        
    return df
```

```python
# CALLING THE API

import os
from time import time
from time import sleep

for i in range(5):
    api_runner()
    print("API Runner Completed.")
    sleep(30)
exit()
```

```python
df5 = df2[["name", "quote.USD.price", "Timestamp"]]
df5 = df5.query("name == 'Bitcoin'")
df5['Time_Only'] = pd.to_datetime(df5['Timestamp']).dt.strftime('%H:%M:%S')
```

```python
#PLOTTING BITCOIN

import matplotlib.ticker as ticker

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(df5['Time_Only'], df5['quote.USD.price'], marker='o', color='orange', linewidth=2)

ax.xaxis.set_major_locator(ticker.MaxNLocator(nbins=5))

plt.xticks(rotation=30, ha='right')

plt.title('Bitcoin (BTC) Price Fluctuation Over Time Tracker', fontsize=14, fontweight='bold')
plt.xlabel('Time Tracker Interval (HH:MM:SS)', fontsize=11)
plt.ylabel('Price in USD ($)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()

plt.show()
```
